## Variables & Data Types

## Operators

```python
a,b=10,3
print(a+b,a-b,a*b,a/b,a//b,a%b,a**b)
print(a>b,a==b,a!=b)
```

## Conditionals

```python
marks=82
if marks>=90:
    print("Grade A")
elif marks>=75:
    print("Grade B")
else:
    print("Grade C")
```

## Loops

```python
for i in range(1,4):
    print("For",i)
i=1
while i<=3:
    print("While",i)
    i+=1
```

## Collections

```python
fruits=["Apple","Banana"]
fruits.append("Mango")
print(fruits)
print((1,2,3))
print({1,2,2,3})
student={"Name":"Prathibha","Age":21}
print(student)
```

## Functions

```python
def greet(name):
    return f"Hello {name}"
print(greet("Prathibha"))
```

## Exception Handling

```python
try:
    print(10/0)
except ZeroDivisionError:
    print("Cannot divide by zero")
```

## File Handling

## OOP

```python
class Student:
    def __init__(self,n,m):
        self.n=n; self.m=m
    def display(self):
        print(self.n,self.m)
Student("Prathibha",95).display()
```

## Pandas

## Mini AI/ML

```python
import pandas as pd
df=pd.DataFrame({"Age":[20,21,22],"Marks":[85,90,95],"Result":["Pass","Pass","Pass"]})
X=df[["Age","Marks"]]
y=df["Result"]
print(X)
print(y)
```
