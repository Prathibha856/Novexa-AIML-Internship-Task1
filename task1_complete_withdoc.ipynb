## 25 Practice Programs

### Factorial

```python
def fact(n):
    if n==0:return 1
    return n*fact(n-1)
print(fact(5))
```

### Palindrome

```python
s="madam"
print(s==s[::-1])
```

### Prime Number

```python
n=29
prime=True
for i in range(2,int(n**0.5)+1):
    if n%i==0:
        prime=False
        break
print(prime)
```

### Largest in List

```python
nums=[5,9,2,11]
print(max(nums))
```
