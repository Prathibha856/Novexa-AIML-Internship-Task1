## Object-Oriented Programming

**Interview Tip:** Understand this concept and be able to explain why it is useful in AI/ML.

```python
class Student:
    def __init__(self,name,marks):
        self.name=name
        self.marks=marks
    def display(self):
        print(self.name,self.marks)

s=Student("Prathibha",95)
s.display()
```

## Mini AI/ML Preprocessing

**Interview Tip:** Understand this concept and be able to explain why it is useful in AI/ML.

```python
df["Result"]=df["Marks"].apply(lambda x:"Pass" if x>=50 else "Fail")
print(df)

# Features (X) and Target (y)
X=df[["Age","Marks"]]
y=df["Result"]

print("Features:")
print(X)
print("Target:")
print(y)
```
